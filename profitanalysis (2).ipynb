```python



table_discount_df = table_df.groupby("Discount")["Profit"].sum().reset_index()


plt.figure(figsize=(10, 5))


colors = ['green' if p > 0 else 'red' for p in table_discount_df["Profit"]]
plt.bar(table_discount_df["Discount"], table_discount_df["Profit"], width=3, color=colors, edgecolor='black', alpha=0.8)

plt.axhline(0, color='black', linewidth=1, linestyle='--') # Zero-profit line
plt.xlabel("Discount Percentage (%)", fontsize=12)
plt.ylabel("Total Profit ($)", fontsize=12)
plt.title("Impact of Discount on Table Profits", fontsize=14, fontweight='bold')
plt.grid(axis='y', linestyle='--', alpha=0.5)

plt.show()
```
